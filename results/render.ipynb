```python
# Cell 1: Load model TD3 từ checkpoint
%cd C:\Users\Admin\Desktop\cs106\doannhom\TD3-master

import numpy as np
import torch
import gymnasium as gym
import TD3

env_name = "Reacher-v5"
seed = 0

# Tạo env để lấy state_dim/action_dim
env = gym.make(env_name, render_mode="rgb_array")
state_dim = env.observation_space.shape[0]
action_dim = env.action_space.shape[0]
max_action = float(env.action_space.high[0])

# Khởi tạo policy và load checkpoint prefix (không thêm _actor/_critic)
policy = TD3.TD3(state_dim, action_dim, max_action)
checkpoint_prefix = r"C:\Users\Admin\Desktop\cs106\doannhom\TD3-master\models\td3/TD3_Reacher-v5_0"   # sẽ load 4 file *_actor, *_critic, *_optimizer
policy.load(checkpoint_prefix)

print("Loaded:", checkpoint_prefix)
```

```python
# Cell 2: Rollout + render animation inline trong notebook
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML

frames = []
obs, _ = env.reset(seed=seed)
done = False
total_reward = 0.0
max_steps = 1000

for _ in range(max_steps):
    action = policy.select_action(np.array(obs))
    obs, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated
    total_reward += reward

    frame = env.render()  # rgb array
    frames.append(frame)

    if done:
        break

env.close()
print("Episode reward:", total_reward, "| frames:", len(frames))

fig = plt.figure(figsize=(8, 6))
im = plt.imshow(frames[0])
plt.axis("off")

def update(i):
    im.set_array(frames[i])
    return [im]

ani = animation.FuncAnimation(fig, update, frames=len(frames), interval=30, blit=True)
plt.close(fig)
HTML(ani.to_jshtml())
```
